# Time dymamics

The Market capitalization growth $\frac{dC_i}{dt} (i \in \{StartUp↗, Star✩, Cash\$Cow\})$ depends on
- company size,
- market share.

## A dependency on company size

The simple formula of a dependency on company size derivers from empirical observations:
 - a "StartUp↗":  if company size = \$1,000 => it growths twice a year = +100%
 - a "Star✩"  : if company size = \\$1,000,000 => it growths on \$100,000 =  +10%
 - a "Cash\\$Cow": if company size = \\$1,000,000,000 => it growths on \$100,000,000 = +1%

so, an annual organic growth = $G_i =\frac{10}{\sqrt[3]{C_i}}$
$$\frac{C_i(t_1)}{C_i(t_o)} \propto [1+ G_i(t_0)]$$

## A dependency on company market share.
$$S_i = \frac{Market\ capitalization\ of\ a\ company}{whole\ market} = \frac{C_i}{M}$$

The closer a company gets to a monopoly position, the faster it tends to grow. The smaller the company, the harder it is to expand..
$$\frac{C_i(t_1)}{C_i(t_o)} \propto [1+ S_i(t_0)]$$

## Static market model at time $t_1$

$$C_i(t_1) = C_i(t_o)*[1+ S_i(t_0)]*[1+G_i(t_0)]$$

$$C_i(t_1) = C_i(t_o)*[1+ \frac{C_i(t_0)}{M(t_0)}]*[1+\frac{10}{\sqrt[3]{C_i(t_0)}}]$$

$$C_i(t_1) = C_i(t_o)*[1+ \frac{C_i(t_0)}{\sum_{j=1}^{\infty} C_j(t_0)}]*[1+\frac{10}{\sqrt[3]{C_i(t_0)}}]$$

# Agents' behaviour

At each step, for example from $t_0$ → $t_1$, there are two subSteps:

1. Each agent redistribute its portfolio
2. Capitalization of each company is changed under new market conditions.

# Market simulation

````python
import numpy as np
N_Steps = 300 # Number of iteration or number of periods.
market_simulation_array = np.empty((arr.shape[0], arr.shape[1], N_Steps - 1), dtype=arr.dtype) # let us to extent the market state at the first step to third dimention = time dimention.
market_simulation_array[:, :, 0] = arr # initial market state at step = 0
````

The inital market state at step = 0

````python
pd.DataFrame(data=market_simulation_array[:, :, 0],index=['MorganStanley', 'WarrenBuffett', 'MS', 'DB', 'C = column Σ'] ,columns=['StartUp↗', "Star✩", 'Cash$Cow', 'capital of the actor = row Σ'])
````

The function
```
market_capitalisation_performance(companies_capitalization: 'np.ndarray') -> 'np.ndarray':
```
calculates a growght of a market capitalization from current step to nect step.

````python
import numpy as np


def market_capitalisation_performance(companies_capitalization: 'np.ndarray') -> 'np.ndarray':
    """
    Calculate the market capitalization performance based on the current market state.

    Parameters:
    current_market_state (np.ndarray): The current state of the market, including trader portfolios and companies capitalization.

    Returns:
    np.ndarray: An array representing the market capitalization performance for each company.
    """


    total_market_capitalization = np.sum(companies_capitalization)

    # Calculate market shares
    market_shares = companies_capitalization / total_market_capitalization

    company_growth_factor = 1 + 10 / (companies_capitalization ** (1 / 3))/10

    # Calculate market capitalization performance
    market_capitalization_performance = (1 + market_shares/10) * company_growth_factor

    return market_capitalization_performance

def trader_portfolios(market_capitalization_performance: 'np.ndarray',
                      trader_portfolios: 'np.ndarray') -> 'np.ndarray':
    """
    Redistribute each trader's capital to the best performing company.

    Parameters:
    market_capitalization_performance (np.ndarray): The performance of each company's market capitalization.
    trader_portfolios (np.ndarray): The current portfolios of traders.

    Returns:
    np.ndarray: Updated trader portfolios after redistribution.
    """

    # Find the indexes of the companies with the biggest performance
    best_companies_indexes: list = np.argwhere(market_capitalization_performance == np.max(market_capitalization_performance)).flatten().tolist()

    # Redistribute each trader's capital to the best performing company
    new_trader_portfolios = np.ones_like(trader_portfolios)
    trader_capitals_except_reminder = trader_portfolios.sum(axis=1) - (3-len(best_companies_indexes))
    trader_capitals_except_reminder = trader_capitals_except_reminder.reshape(-1, 1)  # Reshape to allow broadcasting
    # Each trader invests all their capital in the best performing companies.
    new_trader_portfolios[:, best_companies_indexes] = trader_capitals_except_reminder  // len(best_companies_indexes)

    return new_trader_portfolios
````

````python
arr = np.array(object=[[100, 200, 300],
                       [400, 500, 600],
                       [700, 800, 900],
                       [1000, 1100, 1200]
                       ],
               dtype=np.int64)
N_Steps = 300
market_simulation_array = np.empty((arr.shape[0], arr.shape[1], N_Steps), dtype=arr.dtype)
market_simulation_array[:, :, 0] = arr

# This code simulates a market with economic agents (traders) and companies.
for step in range(N_Steps-1):
    current_market_state = market_simulation_array[:, :, step]

    # Block 1. Calculation a potential market capitalization performance, based on the current companies' capitalization.
    companies_capitalization = current_market_state.sum(axis=0) # sum of each column.
    initial_market_capitalization_performance: np.ndarray = market_capitalisation_performance(companies_capitalization)

    # Block 2.Redistribute each trader's capital to the best performing company. Or use any other strategy.
    current_trader_portfolios = current_market_state
    new_trader_portfolios = trader_portfolios(initial_market_capitalization_performance, current_trader_portfolios)

    # Block 3. Calculate the new market capitalization performance based on the new trader portfolios.
    new_companies_capitalization = new_trader_portfolios.sum(axis=0)
    actual_market_capitalization_performance: np.ndarray = market_capitalisation_performance(new_companies_capitalization)

    # Block 4. Update the market state taking into account the new trader portfolios and the new market capitalization performance.
    next_market_state = new_trader_portfolios * actual_market_capitalization_performance

    # Block 5. Introduce randomness in the total market capitalization while preserving relative shares within each company.
    total_market_capitalization = np.sum(next_market_state)
    std_dev = total_market_capitalization / 3
    random_factor = np.random.normal(loc=total_market_capitalization, scale=std_dev)
    # Ensure random_factor is not negative
    random_factor = max(0.1, random_factor)

    scaling_factor = random_factor / total_market_capitalization
    next_market_state = next_market_state * scaling_factor

    market_simulation_array[:, :, step + 1] = next_market_state
````
